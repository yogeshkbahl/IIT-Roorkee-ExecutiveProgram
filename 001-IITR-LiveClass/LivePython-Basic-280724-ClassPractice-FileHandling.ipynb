```python
## Setup: download the text files this notebook reads (Colab starts with an empty folder)
import urllib.request
for name in ["File1.txt", "File2.txt"]:
    urllib.request.urlretrieve("https://raw.githubusercontent.com/yogeshkbahl/IIT-Roorkee-ExecutiveProgram/main/001-IITR-LiveClass/" + name, name)
```

```python
#Different Different Reading Functionality
```

```python
f=open('File1.txt',"r")
#Reading whole file
print(f.read())
```

```python
f=open('File1.txt',"r")
#Reading first 10 chars
print(f.read(10))
```

```python
f=open('File1.txt',"r")
#Reading each line by line
#print('Yogesh '+f.readline())
print(f.readline())

for x in f:
    print(x)
```

```python
#Closing the File
f.close()
```

```python
#Writing to the file
f=open('File2.txt','at')
f.write('\nEight')
f.write('\n')
f.write('Nine')
f.close()
```

```python
#Read and Write to new file
f2=open('File2.txt','rt')
f3=open('File3.txt','wt')
for x in f2:
    print(x)
    f3.write(x)

f3.close()
f2.close()
```

```python
#Making file Empty
f3=open('File3.txt','wt')
f3.close()
```
