<h3>Join Sets <br>
Join Sets
There are several ways to join two or more sets in Python.

The union() and update() methods joins all items from both sets.

The intersection() method keeps ONLY the duplicates.

The difference() method keeps the items from the first set that are not in the other set(s).

The symmetric_difference() method keeps all items EXCEPT the duplicates.

Note: Both union() and update() will exclude any duplicate items.
</h3>

```python
#Union
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1.union(set2)
print(set3)

set4 = set1 | set2
print(set4)
```

```python
#Join Multiple Sets
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3={True,False,11,12}
set4={"apple","banana","cherry"}
myset=set1|set2|set3|set4
print(myset)

myset1=set1.union(set2,set3,set4)
print(myset1)
```

```python
#Joining a Set and Tuple
set11={11,12,13,14}
tup1=(11,12,3,4)
set11=set11.union(tup1)
print(set11)
```

```python
#Intersection of two sets returns the common 
set1={1,2,3,5,6}
set2={1,2,6,4,7}
set3=set1.intersection(set2)
print(set3)

set4=set1&set2
print(set4)
```

```python
#Differences method will return a new set that will contain only the items from the first set that are not present in the other set
set21 = {"apple", "banana", "cherry"}
set22 = {"google", "microsoft", "apple"}

set23=set21.difference(set22)
print(set23)

set24=set21-set22 # this will allow only set minus set vs difference allows tuple, list, dictionary
print(set24)
```

```python
#The symmetric_difference() method will keep only the elements that are NOT present in both sets.
set25=set21.symmetric_difference(set22)
print(set25)

#Note: The ^ operator only allows you to join sets with sets, and not with other data types like you can with the symmetric_difference() method.
set26 = set21 ^ set22 #You can use the ^ operator instead of the symmetric_difference() method, and you will get the same result.
print(set26)
```
