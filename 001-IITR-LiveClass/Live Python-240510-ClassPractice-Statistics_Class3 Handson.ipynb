```python
#Using CARS.csv
#Reading File Second Way
df=pd.read_csv('C:/Users/yogeshb/001-IITR-LiveClass/IITR Documentation/Dataset/cars.csv')
df
```

```python
df["mpg"].max()
```

```python
#Range
range=df["mpg"].max() - df["mpg"].min()
range
```

```python
#Percentiles 
#90 Percentile value falls under or equal to 243.5
df["hp"].quantile(0.9)
```

```python
#So out of 31 cars 4 cars whose value falls above 90 Percentile
df[df["hp"]>243.50000000000003]
```

```python
#Quartiles 25%, 50%, 75%
Q1=df["hp"].quantile(0.25)
Q2=df["hp"].quantile(0.50)
Q3=df["hp"].quantile(0.75)
print(Q1,Q2,Q3)
```

```python
from scipy.stats import percentileofscore
print(percentileofscore(df["hp"],180.0),
percentileofscore(df["hp"],123.0),
percentileofscore(df["hp"],96.5))
```

```python
#InterQuartile Range
IQR = Q3 - Q1
print(IQR)
```

```python
plt.boxplot(df["hp"])
plt.show()
```

```python
#Upper Limit & Lower Limit explained with Line in BoxPlot
lower_limit=Q1 - 1.5*IQR
upper_limit=Q3 + 1.5*IQR
print(upper_limit,lower_limit)
```

```python
#Variance
df["hp"].var()
```

```python
v=df["hp"].var()

#Standard Deviation calculated using Numpy or Dataframe method
#Numpy Sqrt First Method
np.sqrt(v)
```

```python
#Standard Deviation Function Call second method
df["hp"].std()
```
