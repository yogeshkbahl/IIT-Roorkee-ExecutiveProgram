```python
#Question Create a 2x2, 3x3, 4x4 Identity Matrix using Numpy
from numpy import random

print('Array of 2 Rows and 2 Elements range between 1-100')
x = random.randint(100, size=(2, 2))
print(x)
```

```python
#Generating By Default Identity Matrix 2x2 
x=np.identity(2)
print(x)
print('Data Type=>',x.dtype)
print('Dimension=>',x.ndim)
print('Shape    =>',x.shape)

x=np.identity(3,dtype=int)
print(x)
print('Data Type=>',x.dtype)
print('Dimension=>',x.ndim)
print('Shape    =>',x.shape)
```

```python
#Question Create a Numpy Array of 2x5 with all elements as Zero
y=np.zeros((2,5))
print(y)

y=np.zeros((2,5),dtype=int)
print(y)
```

```python
#Creating Numpy Array of 1s then multiple by 5
y=np.ones((3,5),dtype=int)
print(y)
print(y*5)
```

```python
a = np.arange(12)                    # 1d array
print(a)
print('Single Dim Array')
print('Data Type=>',a.dtype)
print('Dimension=>',a.ndim)
print('Shape    =>',a.shape)
print('#################################')

print('Two Dim Array')
a = np.arange(12).reshape(4,3)                    # 1d array
print(a)
print('Data Type=>',a.dtype)
print('Dimension=>',a.ndim)
print('Shape    =>',a.shape)
print('#################################')

print('Two Dim Array')
a = np.arange(12).reshape(2,6)                    # 1d array
print(a)
print('Data Type=>',a.dtype)
print('Dimension=>',a.ndim)
print('Shape    =>',a.shape)
print('#################################')

print('Two Dim Array')
a = np.arange(12).reshape(4,3)                    # 1d array
print(a)
print('Data Type=>',a.dtype)
print('Dimension=>',a.ndim)
print('Shape    =>',a.shape)
print('#################################')

print('Three Dim Array')
a = np.arange(12).reshape(2,3,2)                    # 1d array
print(a)
print('Data Type=>',a.dtype)
print('Dimension=>',a.ndim)
print('Shape    =>',a.shape)
print('#################################')
```

<h3>Question Create a random array, with the range 0-1000, with a size 50.
    Step 2 Perform the following operations on the same
     1. Print the first ten elements of the array
     2. print the last ten elements of the array
     3. print the elements from index 10:25
     4. print the element at index 22
     5. print the array using negative indexing
     6. print the last ten elements of the array using negative indexing
</h3>

```python
import numpy as np
x = np.random.randint(0,1000, size=50)
print(x)
```

```python
print(x[0:10]) #1. First 10 Elements
print(x[-10:]) #2. Last 10 Elements
print(x[10:26]) #3. Print the element 10:25
print(x[22])    #4 print the element 22
print(x[::-1])
print(x[-1:-11:-1])
```
