Python Exceptional Handling-

How many types of errors-

SyntaxError:
This happens when there's a mistake in the way the code is written, like using the wrong word or forgetting a punctuation mark.

TypeError:
This occurs when you try to do an action with the wrong type of thing, like trying to add a word to a number.

NameError:
This occurs when you try to use a name (like a variable or function) that hasn't been defined yet.

IndexError:
This happens when you're trying to find something in a list, but you ask for an item that's not there.

KeyError:
This occurs when you're using a dictionary and ask for information using a key that doesn't exist.

ValueError:
This occurs when you give a function the wrong kind of information, like trying to tell it your age using words instead of a number.

AttributeError:
This occurs when you try to get information from an object (like a variable or function), but that object doesn't have what you're looking for.

IOError:
This happens when you're trying to read or write a file, but something goes wrong, like the file not being where you said it would be.

ZeroDivisionError:
This occurs when you're trying to divide a number by zero, which is impossible in mathematics.

ImportError:
This occurs when you try to bring in code from somewhere else, but Python can't find or use that code.
