```python
A survey was conducted in 2019 for all plumbers where the average salary of the plumber was found to be 
hundred dollars while taking a sample of 30 plumbers 
for a new survey. It was found out that the average salary of the plumber was one $140 with a standard deviation 
of $20. Calculate this with a confident of 95%.
```

```python
from scipy import stats   #sci compy 
import numpy as np
```

```python
#data
sample_mean = 140
population_mean = 100
n = 30
sample_std = 20
alpha = 0.05


t_statistic = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
print(f"t_statistic : {t_statistic}")

degree_of_freedom = n -1
print(f"degree_of_freedom : {degree_of_freedom}")

#table   ppf - percent point function
t_critical_value = stats.t.ppf(1 - alpha/2, degree_of_freedom)
print(f"t_critical_value : {t_critical_value}")

p_value = 2 * (1 - stats.t.cdf(abs(t_statistic), degree_of_freedom))
print(f"p_value:{p_value}")

#universal 
if p_value < alpha:
    print("Reject h0")
else:
    print("Accept h0")
```

```python
Factory shampoo, (gAURAV)
A factory has a machine that dispenses 80 mL of shampoo in a bottle and an employee believes 
that the average amount  40 of shampoo for that he measured but 78ML with a standard deviation of 2.5 ML.
Can you please state your null and alternate hypothesis and also provide in a evidence to support the idea that 
the machine is working properly or is the machine not working properly at 95% CI
```

```python
Null : (avg amount of shampoo = 80) - machine is working fine
Alternate : (avg amount of shampoo != 80) - machine is not working fine  
    
(more, less) - two tailed z test
```

```python
import numpy as np
from scipy.stats import norm

#degree of freedom > z stats does not has degree of freedom

#data
sample_mean = 78
population_mean = 80
std_Dev = 2.5
n=40
alpha = 0.05   #95 

z_statistic = (sample_mean - population_mean) / (std_Dev/np.sqrt(n))
print(f"z_statistic : {z_statistic}")

#coding - critical value? no
critical_z = norm.ppf(1 - alpha/2)  #why /2. how main tailed? 2  one
print(f"critical_z : {critical_z}")

p_value = 2 * (1 - norm.cdf(abs(z_statistic)))
print(f"p_value  : {p_value}")

if p_value < alpha:
    print("Reject h0 : (avg amount of shampoo != 80) - machine is not working fine ")
else:
    print("Accept h0: (avg amount of shampoo = 80) - machine is working fine")
```

```python
In a sample survey conducted in Bangalore, it was found that 960 people out of 1860 people were vegetarian and the others were non-vegetarian. 
Can you test that in Bangalore vegetarian and non-vegetarian are 50% of each other or equal 
to each other in population? Can you also do this exercise at 1% level of significance?   #99%
```

```python
null : veg and non-veg 50-50
alter : veg and non-veg are not 50-50
```

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest  #code
```

```python
vegetarian_count = 960
total_population = 1860

p_hat = vegetarian_count / total_population
print(f"p_hat:{p_hat}")

alpha = 0.01

# nobs = numbers of observation

z_stat, p_value = proportions_ztest(count=vegetarian_count,
                                   nobs = total_population,
                                   value=0.50,
                                   alternative="two-sided") #two tailed test

print(f"z_stat:{z_stat}")
print(f"p_value:{p_value}")

if p_value < alpha:
    print("Reject h0 : veg and non-veg are not 50-50  ")
else:
    print("Accept h0: veg and non-veg 50-50 ")
```

```python
import numpy as np
from scipy.stats import chi2

observed = np.array([45, 43, 32, 45, 46])  #ashish
expected = np.array([20, 20, 20, 20, 20])

chi2_statistic = np.sum ((observed - expected) **2/expected)
print(f"chi2_statistic: {chi2_statistic}")


#critical value - table 
df=4
alpha = 0.05
critical_value = chi2.ppf(1-alpha,df)
print(f"critical_value : {critical_value}")

p_value = 1 - chi2.cdf(chi2_statistic,df)

if p_value < alpha:
    print("Reject h0 : ex != ob  ((do not buy the rest)) ")
else:
    print("Accept h0:  ex = ob  (buy the rest) ")
```

```python
from scipy.stats import f_oneway
 #mean 
alpha = 0.05

fertilizer_a = [12, 13, 15, 14, 16, 13, 15, 14, 15, 13]
fertilizer_b = [14, 15, 16, 15, 17, 15, 16, 14, 16 , 15]
fertilizer_c = [11, 12, 13, 13, 14, 13, 13, 11, 13 , 12]

# len(fertilizer_a) == len(fertilizer_b) == len(fertilizer_c)
```

```python
null : fertilizer_a = fertilizer_b = fertilizer_c
alternate : fertilizer_a != fertilizer_b != fertilizer_c
```

```python
f_statistic, p_value = f_oneway(fertilizer_a, fertilizer_b,fertilizer_c)
print(f"f_statistic: {f_statistic}")
print(f"p_value  :{p_value}")
```

```python
if p_value < alpha:
    print("Reject h0 : fertilizer_a != fertilizer_b != fertilizer_c")
else:
    print("Accept h0:  fertilizer_a = fertilizer_b = fertilizer_c ")
```
