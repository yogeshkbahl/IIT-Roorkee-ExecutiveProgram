## How to create a basic graph on Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
year = [2019,2020,2021,2022,2023,2024]
population = [1.2,3.4,5,6,5.2,8]
population_ind = [1.5,3.5,5.1,7.2,9,10]
```

```python
fig, (axes1, axes2) = plt.subplots(1,2,figsize=(10,5)) #figsize we can control the size of the canavas
axes1.plot(year,population,color="red")
axes1.set_title("Population Growth of China")
axes1.set_xlabel("Year")
axes1.set_ylabel("Population")

axes2.bar(year,population_ind,color="green")
axes2.set_title("Population Growth of India")
axes2.set_xlabel("Year")
axes2.set_ylabel("Population")
plt.show()
```

```python
fig, axes = plt.subplots(figsize=(10,5)) #figsize we can control the size of the canavas
axes.plot(year,population,color="red")
axes.bar(year,population_ind,color="green")
axes.set_title("Population Growth of India vs China")
axes.set_xlabel("Year")
axes.set_ylabel("Population")
```

## Scatter Plot
It used to compare relationship between variables

```python
import pandas as pd
```

```python
df = pd.read_csv("bike-sharing-demand.csv")
df.sample(5)
```

```python
x= df["temp"].to_numpy()
y= df["humidity"].to_numpy()
```

```python
fig, ax = plt.subplots()
ax.scatter(df[df["season"]==1]["temp"],df[df["season"]==1]["atemp"],label="spring")
ax.scatter(df[df["season"]==2]["temp"],df[df["season"]==2]["atemp"],label="summer")
ax.scatter(df[df["season"]==3]["temp"],df[df["season"]==3]["atemp"],label="fall")
ax.scatter(df[df["season"]==4]["temp"],df[df["season"]==4]["atemp"],label="winter")
ax.set_title("Relationship between Temperature and Feel like temperature")
ax.set_xlabel("Temperature")
ax.set_ylabel("Feel Like Temp")
ax.legend()
```

## Steps to find Interquartile range

* **Step 1**: Put the numbers in order
1, 2, 5, 6, 7, 9, 12, 15, 18, 19, 27.
* **Step 2**: Find the median.
1, 2, 5, 6, 7, 9, 12, 15, 18, 19, 27.

* **Step 3**: Place parentheses around the numbers above and below the median.
Not necessary statistically, but it makes Q1 and Q3 easier to spot.
(1, 2, 5, 6, 7), 9, (12, 15, 18, 19, 27).
* **Step 4**: Find Q1 and Q3
Think of Q1 as a median in the lower half of the data and think of Q3 as a median for the upper half of data.
(1, 2, 5, 6, 7),  9, ( 12, 15, 18, 19, 27). Q1 = 5 and Q3 = 18.
* **Step 5**: Subtract Q1 from Q3 to find the interquartile range.
18 – 5 = 13.
