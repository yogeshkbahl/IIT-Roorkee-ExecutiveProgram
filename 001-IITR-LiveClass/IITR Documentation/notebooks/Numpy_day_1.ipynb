```python
## adding 2 numpy array

arr1 = np.array([[1,5],
                 [3,4]])

arr2 = np.array([[2,2],
                 [2,2]])
```

```python
arr1+arr2 #elementwise addition
```

```python
arr1 = np.array([[1,2],
                 [3,4]])

arr2 = np.array([[2,2],
                 [2,2]])
```

```python
arr1*arr2 #elemetwise mutliplication
```

```python
arr1 = np.array([[4,9],
                 [16,25]])

np.sqrt(arr1)
```

```python
np.log(arr1)
```

```python
np.sin(arr1)
```

```python
np.sum(arr1) # sum of all values in the array
```

```python
np.sum(arr1,axis=0) #takes summation column wise
```

```python
np.sum(arr1,axis=1) #takes summation row wise
```

## Array Manupulation

```python
arr1 = np.array([[3,2],
                 [-1,4]])
arr2 = np.array([[11,-11],
                 [-11,11]])
```

```python
np.hstack((arr1,arr2)) #horizontal stack
```

```python
np.vstack((arr1,arr2)) # vertical stack
```

```python
arr3 = np.array([[1,2,3],
                 [4,5,6]])

arr4 = np.array([[7,8,9],
                 [10,11,12]])

#concatenate the above numpy arrays vertically and horizonatlly?
```

```python
np.hstack((arr3,arr4))
```

```python
np.vstack((arr3,arr4))
```
