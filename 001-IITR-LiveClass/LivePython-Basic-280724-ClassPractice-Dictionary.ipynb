<h2>Python Dictionary </h2>
<h3> Dictionaries are used to store data values in key:value pairs.

A dictionary is a collection which is ordered*, changeable and do not allow duplicate

Dictionaries are written with curly brackets, and have keys and values:
</h3>s.

```python
#Dictionary Created
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,       #Duplicates not allowed
  "year": 2020,       #Duplicates not allowed  
  "year": 2024
}
print(thisdict)
print(thisdict["brand"])
print(thisdict["year"])

print(len(thisdict))

thisdict1 = {
  "brand": "Ford",
  "electric": False,
  "year": 1964,
  "colors": ["red", "white", "blue"]
}

print(thisdict1)
print(len(thisdict1))
print(type(thisdict1))

dict2=dict()
print(dict2)
print(len(dict2))
```

<h3> Access Dictionary Items</h3>

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
x = thisdict["model"]

y = thisdict.get("model")
print(x)
print(y)

z = thisdict.keys()
z1=thisdict.values()
print(z)
print(z1)
```

```python
#Adding new key
thisdict["color"]="Red"
print(thisdict)

z = thisdict.keys()
z1=thisdict.values()
print(z)
print(z1)

print(thisdict.items())
print('--------Break---------')
print(thisdict.keys())
print(thisdict.values())
print(thisdict.items())

if "color" in thisdict:
    print("Yes the color is one of the keys in the thisdict")

#Update or Overridding current value
thisdict["year"]=2024
print(thisdict)

thisdict.update({"color":'blue'})
print(thisdict)
```

```python
#Removing dictionary items
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "color":'Grey'
}
thisdict.pop("model")
print(thisdict)

thisdict.popitem()
print(thisdict)

del thisdict["brand"]
print(thisdict)

thisdict.clear()
print(thisdict)
```

```python
#Looping Dictionary 
thisdict1 = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "color":'blue'
}

print(thisdict1)

for x in thisdict1:
    print("Key =>"+x)
print('--------Break---------')
for x in thisdict1:
    print("Value =>"+str(thisdict1[x]))
print('--------Break---------')
for x in thisdict1.values():
  print(x)
print('--------Break---------')    
for x in thisdict1.keys():
  print(x)
print('--------Break---------')    
for x, y in thisdict1.items():
  print(x, y)
```

```python
#Copy Dictionary
mydict = thisdict1.copy()
print(mydict)

mydict1=dict(thisdict1)
print(mydict1)
```

```python
#Nested Dictionary
myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007,
    "age"  : 18
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011
  }
}

print(myfamily)
```

```python
#2nd Nested Example
child1 = {
  "name" : "Emil",
  "year" : 2004
}
child2 = {
  "name" : "Tobias",
  "year" : 2007
}
child3 = {
  "name" : "Linus",
  "year" : 2011
}

myfamily = {
  "child1" : child1,
  "child2" : child2,
  "child3" : child3
}

print(myfamily)
print(myfamily["child1"])
print(myfamily["child1"]['name'])




for x,obj in myfamily.items():
    print(x)
    for y in obj:
        print(y+':',obj[y])
```

```python
#Clear the Dictionary Object
myfamily.clear()
print(myfamily)
```
