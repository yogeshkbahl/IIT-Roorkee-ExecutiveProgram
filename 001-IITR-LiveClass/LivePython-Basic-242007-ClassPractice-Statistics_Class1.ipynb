<h3>- Standard Variance </h3>
	- Standard Variance talks about how the data points are spread from the mean
	- Low Variance values means the data are close to each other like [12,13,10,9,19,16,15]
	- High Variance values means the data are far away from each other [100,999, 1,500,15h3>

```python
#Variance or Standard Variance
import numpy as np
import statistics as stats
x=[12,13,14,15,9,10,17,19]
print('Low Variance=>',stats.variance(x))
print(np.mean(x))

y=[100,999,32453675,1,4,67,345654745]
print('High Variance =>',stats.variance(y))
print(np.mean(y))


```

#Deviation or Standard Deviation
<h3>- Standard Deviation </h3>
	- will talk about the magnitude by which each data points are separated
	- This is also used to express the magnitude by which the members of the group differ from the mean value for the group

```python
import numpy as np
import statistics as stats
x=[12,13,14,15,9,10,17,19]
print('Low Variance =>',stats.variance(x))
print('Mean=>',np.mean(x))
print('Standard Deviation',stats.stdev(x))

print('----------------')

y=[100,999,32453675,1,4,67,345654745]
print('High Variance =>',stats.variance(y))
print('Mean=>',np.mean(y))
print('Standard Deviation',stats.stdev(y))
```

```python
age = [-190,250,260,1,21,31,41,45,50,51,59,60,70,80,90,85,76,46,34,11,-160,1]

import matplotlib.pyplot as plt
import numpy as np
plt.boxplot(age)
plt.show()

```

```python
#Quartile 1
Q1=np.percentile(age,25)
Q1
```

```python
#Quartile 3
Q3=np.percentile(age,75)
Q3
```

```python
#InterQuartile Range
IQR = Q3 - Q1
IQR
```

```python
#Upper Threshold
UT = Q3+1.5*IQR
UT
```

```python
#Lower Threshold
LT = Q3 - 1.5*IQR
LT
```

```python
# Removing Outliers from the Set
#Before Running
age
```

```python
for i in age:
    if i>UT or i<LT:
        age.remove(i)

#After running the age and removing Outliers
age
```
