```python
#Filtering data in Python
#Single Filter Condition
df[df["casual"]<5]

#Multiple Condition using AND
#df[(df["casual"]<5) & (df["humidity"]>79)]


#Multiple Condition using AND and OR Combination
df[((df["casual"]<5) & ((df["humidity"]>79) | (df["registered"]<11))) & (df["count"]==1)]
```

```python
tot=df[((df["casual"]<5) & ((df["humidity"]>79) | (df["registered"]<11))) & (df["count"]==1)]["count"].sum()
print(f"Total rows return is {tot}")
```
